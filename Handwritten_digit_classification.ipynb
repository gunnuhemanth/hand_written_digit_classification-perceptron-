```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.layers import Dense, Flatten
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
```

```python
#this make convergence fast
```

```python
x_train=x_train/255
```

```python
x_test=x_test/255
```

```python
x_train[0]
```

```python
from tensorflow.keras import Sequential
model=Sequential()
```

```python
model.add(Flatten(input_shape=(28, 28)))
# this flattens our image data 28 * 28 to 1 * 784
```

```python
model.add(Dense(128, activation='relu'))
#we are keeping 128 nodes, and we dont need to mention that we have 784 inputs(it will be automatically figured out)
```

```python
model.add(Dense(10, activation='softmax'))
#since we have >1 nodes in output always use softmax
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy', optimizer='Adam')
#this is for checking how much loss while we check output probabilities and actual data differ
# For example:

# If the true label = class 2

# The softmax output = [0.1, 0.1, 0.8]

# The loss will be very small (because the model predicted correctly).

# But if softmax output = [0.6, 0.3, 0.1] → the loss will be large, meaning the model did badly.
```

```python
history=model.fit(x_train, y_train, epochs=10, validation_split=0.2)
```

```python
y_prob=model.predict(x_test)
```

```python
y_pred=y_prob.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test, y_pred)
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.legend(['loss', 'val_loss'])
plt.show()
```
